```python
# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Step 2: Import Dataset
df=pd.read_csv('/Salary_dataset.csv')
```

```python
df.describe()
```

```python
df.info()
```

```python
# Step 3: Data Visualization
plt.scatter(data=df,x='YearsExperience',y='Salary',s=200)
plt.title("salary based on years of experince")
plt.xlabel("years od experince")
plt.ylabel("salary")
plt.show()
```

```python
# Step 4: Separate Independent and Dependent Variables

X = df[['YearsExperience']].values
X=df[['YearsExperience']].values
y=df['Salary'].values
```

```python
# Step 5: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(

    X, y, test_size=0.2, random_state=1

)
```

```python
# Step 6: Model Training
salary_model = LinearRegression()

salary_model.fit(X_train, y_train)
```

```python
# Step 7: Find Coefficient and Intercept
print("Coefficient:",salary_model.coef_)
print("intercept:",salary_model.intercept_)
```

```python
# Step 8: Prediction
y_pred=salary_model.predict(X_test)
print("Actual:",y_test)
print("Predicted:",y_pred)
```

```python
# Step 9: Model Evaluation
print("R2 Score:", r2_score(y_test, y_pred))

print("MSE:", mean_squared_error(y_test, y_pred))
```

```python
# Step 10: Visualization

plt.scatter(X_test, y_test, color='red', label='Actual')

plt.scatter(X_test, y_pred, color='green', label='Predicted')

plt.xlabel("Years of Experience")

plt.ylabel("Salary")

plt.legend()

plt.show()
```
